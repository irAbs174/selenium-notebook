# 🕸️ Selenium в Google Colab: Готовый шаблон

Готовый шаблон для запуска Headless Chrome с Selenium внутри Google Colab. Эта настройка автоматически управляет версиями драйвера и включает обходные методы для типичных систем антибот-защиты.

## 🚀 Возможности

* **Никаких ручных обновлений:** Используется `webdriver-manager` для автоматической загрузки подходящего бинарного файла ChromeDriver.
* **Готово для Colab:** Предварительно настроено с обязательными флагами `--no-sandbox` и `--disable-dev-shm-usage`, необходимыми для запуска Chrome в контейнерах Colab.
* **Обход антибот-систем:** Маскирует среду автоматизации, отключая функции Blink `AutomationControlled` и подменяя User-Agent на стандартный Windows.
* **Визуальная отладка:** Включён фрагмент кода для отображения скриншотов прямо в ноутбуке Colab.

## 🛠️ Использование

### 1. Установка зависимостей

### 2. Основная настройка

Скопируйте и вставьте этот скрипт, чтобы инициализировать драйвер, обойти ботов и сделать скриншот для проверки.

````
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
import subprocess

# Автоматическая установка webdriver-manager, если он отсутствует
try:
    from webdriver_manager.chrome import ChromeDriverManager
except ImportError:
    print("webdriver-manager не найден, устанавливаем...")
    subprocess.check_call(['pip', 'install', 'webdriver-manager'])
    from webdriver_manager.chrome import ChromeDriverManager

def get_colab_driver():
    options = webdriver.ChromeOptions()

    # 1. Флаги для headless-режима (критически важны для Linux-окружения Colab)
    options.add_argument('--headless=new')
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-dev-shm-usage')

    # 2. Отключение триггеров обнаружения ботов
    options.add_argument('--disable-blink-features=AutomationControlled')
    options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36")

    # 3. Автоматическое управление бинарным файлом ChromeDriver
    service = Service(ChromeDriverManager().install())
    return webdriver.Chrome(service=service, options=options)

# Инициализация
driver = get_colab_driver()

# Проверка настройки
driver.get("https://example.com")
driver.save_screenshot("visual_check.png")

print("Заголовок страницы:", driver.title)
````

### 3. Просмотр результата

Так как браузер работает в headless-режиме, используйте IPython, чтобы увидеть, что видит браузер:

````
from IPython.display import Image, display
display(Image("visual_check.png"))
````

### 4. Очистка

Всегда закрывайте драйвер, чтобы освободить память в вашем экземпляре Colab:

````
driver.quit()
````
