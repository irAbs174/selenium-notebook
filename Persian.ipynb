# 🕸️ Selenium در Google Colab: قالب آماده و نهایی

یک قالب آماده و قابل اجرا برای راه‌اندازی مرورگر Headless Chrome به همراه Selenium در محیط Google Colab. این تنظیمات به‌طور خودکار نسخه‌های درایور را مدیریت کرده و شامل روش‌های ضروری برای دور زدن سیستم‌های رایج تشخیص ربات است.

## 🚀 قابلیت‌ها

* **بدون نیاز به به‌روزرسانی دستی:** با استفاده از `webdriver-manager`، باینری مناسب ChromeDriver به‌طور خودکار دریافت می‌شود.
* **سازگار با Colab:** از قبل با پرچم‌های اجباری `--no-sandbox` و `--disable-dev-shm-usage` پیکربندی شده است که برای اجرای Chrome در کانتینرهای Colab ضروری هستند.
* **دور زدن تشخیص ربات:** با غیرفعال کردن ویژگی‌های Blink مربوط به `AutomationControlled` و جعل User-Agent استاندارد ویندوز، محیط اتوماسیون را پنهان می‌کند.
* **اشکال‌زدایی تصویری:** شامل یک قطعه کد برای نمایش اسکرین‌شات‌های گرفته‌شده در حالت بدون رابط، مستقیماً داخل دفترچه Colab.

### 2. راه‌اندازی اصلی

این اسکریپت را کپی و پیست کنید تا درایور شما راه‌اندازی شود، تشخیص ربات دور زده شود و یک اسکرین‌شات از نتیجه گرفته شود.

````
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
import subprocess

# نصب خودکار webdriver-manager در صورت عدم وجود
try:
    from webdriver_manager.chrome import ChromeDriverManager
except ImportError:
    print("webdriver-manager یافت نشد، در حال نصب...")
    subprocess.check_call(['pip', 'install', 'webdriver-manager'])
    from webdriver_manager.chrome import ChromeDriverManager

def get_colab_driver():
    options = webdriver.ChromeOptions()

    # 1. پرچم‌های حالت بدون رابط (برای محیط لینوکس Colab حیاتی هستند)
    options.add_argument('--headless=new')
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-dev-shm-usage')

    # 2. غیرفعال‌سازی نشانگرهای تشخیص ربات
    options.add_argument('--disable-blink-features=AutomationControlled')
    options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36")

    # 3. مدیریت خودکار باینری ChromeDriver
    service = Service(ChromeDriverManager().install())
    return webdriver.Chrome(service=service, options=options)

# راه‌اندازی
driver = get_colab_driver()

# آزمایش تنظیمات
driver.get("https://example.com")
driver.save_screenshot("visual_check.png")

print("عنوان صفحه:", driver.title)
````
